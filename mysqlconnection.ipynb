```python
import pandas as pd
from getpass import getpass
from urllib.parse import quote_plus
from sqlalchemy import create_engine, text

emp = pd.read_csv("clean/employee_final.csv", parse_dates=["StartDate", "ExitDate"])

pwd = quote_plus(getpass("MySQL password: "))
engine = create_engine(f"mysql+pymysql://root:{pwd}@localhost:3306/hr_project")
emp.to_sql("employees", engine, if_exists="replace", index=False, chunksize=500)

with engine.connect() as con:
    print("Rows loaded:", con.execute(text("SELECT COUNT(*) FROM employees")).scalar())
```
